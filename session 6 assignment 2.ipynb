```python
#4.Simulate a Zomato-style restaurant ratings dataset with some missing ratings. Use forward fill (method='ffill') to fill missing values in the ratings column, then use backward fill (method='bfill') for any remaining missing values. 
#Show the before and after results.
import pandas as pd

df = pd.read_csv("zomato_ratings.csv")

print("Before:")
print(df)

df["Rating"] = df["Rating"].ffill()
df["Rating"] = df["Rating"].bfill()

print("\nAfter:")
print(df)
```
